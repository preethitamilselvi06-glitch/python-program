```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\acer\Downloads\GitHub\python-program\notebooks\clean_transactions.csv")
df.head()
df.shape
df.info()
df.describe()
df.isnull().sum()
df.duplicated().sum()
df = df.drop_duplicates()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['amount'], bins=30)
plt.title("Transaction Amount Distribution")
plt.show()
```

```python
sns.countplot(x='user_city_tier', data=df)
plt.title("User_City_tier Distribution")
plt.show()
```

```python
df['receiver_id'].value_counts().head(10)
```

```python
df['receiver_id'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 receiver")
plt.show()
```

```python
sns.boxplot(x=df['amount'])
plt.title("Transaction Amount Boxplot")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap='RdYlGn')
plt.show()
```
